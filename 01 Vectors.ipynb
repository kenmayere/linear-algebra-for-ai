##### Vectors
In this nb, I will be discussing vectors. Depending on the field of study, this term has different meanings.
    - Mathematics and physics defines vectors as a quantity possing both size and direction, often presented as an arrow in space
    - Biology and medicine defines a vector as an organism that transmits a disease-causing pathogen from another
    - Computer science defines vectors as an ordered array or sequence of values stored with a single index, and
    - Aviation and navigation look at vectors as compass direction or set of instructions used to steer or guide an aircraft or direct engine thrust.
    Interestly, across different fields, there overlaps, for example, I always believe that aviation vectors will outline size(magnitude/altitude/nautical miles) and direction (heading).

I will be moving around the mathematical and computer science perspectives. So vectors will mean values/numbers in space or contained in a certain dimension.

In ML, associated with vectors, we have scalars which are just numbers, matrices and tensors which I will discuss in other nbs.
